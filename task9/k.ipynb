```python
# =============== 导入库 ===============
import pandas as pd
import numpy as np
import os
import mplfinance as mpf
import matplotlib.pyplot as plt
from datetime import datetime
import warnings
warnings.filterwarnings('ignore')

# ===================== 配置参数 =====================
# 数据文件在task8，输出保存到task9
DATA_DIR = r"D:\vscode\project\task8\stock_data"  # 数据源在task8
OUTPUT_DIR = r"D:\vscode\project\task9\output_charts"  # 输出到task9
os.makedirs(OUTPUT_DIR, exist_ok=True)  # 确保输出目录存在

# 三只股票
STOCKS = {
    "002460": "赣锋锂业",
    "600733": "北汽蓝谷", 
    "300182": "捷成股份"
}

# 图表配置
CHART_CONFIG = {
    'dpi': 300,                    # 高分辨率
    'figsize': (16, 10),           # 更大的图表尺寸
    'font_size': 12,               # 字体大小
    'grid_alpha': 0.3,             # 网格透明度
    'show_last_n_days': 120        # 默认显示最近120天数据
}

print("="*70)
print("📈 任务9：股票分析系统 - 从task8读取数据，保存到task9")
print("="*70)
print(f"📂 数据源目录: {DATA_DIR}")
print(f"📁 输出目录: {OUTPUT_DIR}")
print("="*70)

def preprocess_data(df):
    """数据预处理：清洗、验证、处理异常值"""
    df = df.copy()
    
    # 1. 基本清洗：去重
    df.drop_duplicates(inplace=True)
    print(f"   - 去重后数据量: {len(df)}")
    
    # 2. 确保数值列是正确的数据类型
    numeric_cols = ['open', 'high', 'low', 'close', 'volume']
    for col in numeric_cols:
        if col in df.columns:
            # 将列转换为浮点类型，避免整数溢出问题
            df[col] = pd.to_numeric(df[col], errors='coerce')
            print(f"   - 列 '{col}' 转换为数值类型，数据类型: {df[col].dtype}")
    
    # 3. 处理缺失值
    for col in numeric_cols:
        if col in df.columns:
            original_missing = df[col].isnull().sum()
            if original_missing > 0:
                print(f"   - 列 '{col}' 原始缺失值: {original_missing}个")
                df[col] = df[col].ffill()  # 前向填充
                df[col] = df[col].bfill()  # 后向填充
                remaining_missing = df[col].isnull().sum()
                print(f"   - 填充后剩余缺失值: {remaining_missing}个")
    
    # 4. 验证数据合理性
    if 'high' in df.columns and 'low' in df.columns and 'close' in df.columns:
        # 检查价格逻辑：最高价 >= 收盘价 >= 最低价
        invalid_mask = (df['high'] < df['close']) | (df['close'] < df['low'])
        if invalid_mask.any():
            count_invalid = invalid_mask.sum()
            print(f"   - ⚠️  发现{count_invalid}条异常价格记录，已修正")
            df.loc[invalid_mask, 'high'] = df.loc[invalid_mask, ['high', 'close']].max(axis=1)
            df.loc[invalid_mask, 'low'] = df.loc[invalid_mask, ['low', 'close']].min(axis=1)
    
    # 5. 处理成交量异常
    if 'volume' in df.columns:
        # 确保成交量非负
        df['volume'] = df['volume'].clip(lower=0)
        
        # 处理异常大的成交量
        if len(df) > 10:
            volume_mean = df['volume'].mean()
            volume_std = df['volume'].std()
            upper_bound = volume_mean + 3 * volume_std
            upper_bound_int = int(round(upper_bound))
            
            extreme_mask = df['volume'] > upper_bound
            extreme_count = extreme_mask.sum()
            if extreme_count > 0:
                print(f"   - 修正{extreme_count}个异常大成交量（阈值: {upper_bound_int:,}）")
                df.loc[extreme_mask, 'volume'] = upper_bound_int
    
    return df

def calculate_technical_indicators(df):
    """计算技术指标，包含精度控制"""
    df = df.copy()
    
    # 1. 日收益率
    df["日收益率"] = df["Close"].pct_change() * 100
    df["日收益率"] = df["日收益率"].round(6)
    
    # 2. 移动平均线
    df["MA5_SMA"] = df["Close"].rolling(window=5, min_periods=1).mean().round(4)
    df["MA10_SMA"] = df["Close"].rolling(window=10, min_periods=1).mean().round(4)
    
    # 3. 指数移动平均线
    df["MA5_EMA"] = df["Close"].ewm(span=5, adjust=False, min_periods=1).mean().round(4)
    df["MA10_EMA"] = df["Close"].ewm(span=10, adjust=False, min_periods=1).mean().round(4)
    
    # 4. 成交量移动平均
    df["Volume_MA5"] = df["Volume"].rolling(window=5, min_periods=1).mean().round(0).astype(int)
    df["Volume_MA10"] = df["Volume"].rolling(window=10, min_periods=1).mean().round(0).astype(int)
    
    return df

def create_professional_chart(df, name, code, output_path=None):
    """创建专业级K线图"""
    # 1. 数据筛选
    if len(df) > CHART_CONFIG['show_last_n_days']:
        plot_df = df.tail(CHART_CONFIG['show_last_n_days']).copy()
        title_suffix = f" (最近{CHART_CONFIG['show_last_n_days']}天)"
    else:
        plot_df = df.copy()
        title_suffix = ""
    
    # 2. 检查有效数据
    valid_data = plot_df.dropna(subset=['Open', 'High', 'Low', 'Close', 'Volume'])
    if len(valid_data) < 5:
        print(f"   - ⚠️  数据量不足（{len(valid_data)}条），无法生成有意义的图表")
        return
    
    plot_df = valid_data
    
    # 3. 确保绘图数据是正确的数据类型
    for col in ['Open', 'High', 'Low', 'Close', 'Volume']:
        if col in plot_df.columns:
            plot_df[col] = pd.to_numeric(plot_df[col], errors='coerce')
    
    # 4. 创建自定义样式
    market_colors = mpf.make_marketcolors(
        up='red', down='green', 
        edge='inherit', 
        wick={'up': 'red', 'down': 'green'},
        volume={'up': 'red', 'down': 'green'},
        ohlc='inherit'
    )
    
    mpf_style = mpf.make_mpf_style(
        marketcolors=market_colors,
        gridstyle=':',  # 虚线网格
        gridcolor='gray',
        facecolor='white',  # 白色背景
        rc={'font.family': 'SimHei', 'axes.unicode_minus': False},  # 支持中文
        y_on_right=True,
        figcolor='white'
    )
    
    # 5. 添加额外的技术线
    apds = [
        mpf.make_addplot(plot_df['MA5_EMA'], color='blue', width=1.5, label='5日EMA'),
        mpf.make_addplot(plot_df['MA10_EMA'], color='orange', width=1.5, label='10日EMA'),
    ]
    
    # 6. 创建图表
    try:
        print("   - 📊 生成K线图...")
        fig, axes = mpf.plot(
            plot_df,
            type='candle',           # 蜡烛图
            style=mpf_style,
            title=f'{name} ({code}) 任务9分析{title_suffix}',
            ylabel='价格 (元)',
            ylabel_lower='成交量',
            volume=True,             # 显示成交量
            addplot=apds,            # 添加额外指标
            figratio=(16, 10),       # 宽高比
            figscale=1.0,            # 整体缩放
            datetime_format='%Y-%m-%d',
            xrotation=45,           # X轴标签旋转
            tight_layout=True,
            returnfig=True,          # 返回figure对象以便进一步定制
            show_nontrading=False,   # 不显示非交易日
            scale_padding={'left': 0.05, 'right': 0.05, 'top': 0.1, 'bottom': 0.15},
            warn_too_much_data=1000  # 增加数据量警告阈值
        )
        
        # 7. 进一步定制图表
        if axes:
            ax_main = axes[0]  # 主图表（价格）
            ax_main.grid(True, alpha=CHART_CONFIG['grid_alpha'], linestyle='--')
            ax_main.set_title(ax_main.get_title(), fontsize=CHART_CONFIG['font_size'] + 4, fontweight='bold')
            ax_main.set_ylabel(ax_main.get_ylabel(), fontsize=CHART_CONFIG['font_size'] + 2)
        
        print("   - ✅ K线图生成成功")
        
    except Exception as e:
        print(f"   - 📊 图表生成失败: {str(e)}")
        return
    
    # 8. 保存图表
    if output_path:
        try:
            plt.savefig(output_path, dpi=CHART_CONFIG['dpi'], bbox_inches='tight')
            print(f"✅ 图表已保存至: {output_path}")
        except Exception as e:
            print(f"   - 💾 保存图表失败: {str(e)}")
    
    # 9. 显示图表
    try:
        plt.show()
    except Exception as e:
        print(f"   - 👁️  显示图表失败: {str(e)}")
    
    plt.close()

# ===================== 主处理流程 =====================
for code, name in STOCKS.items():
    print(f"\n{'='*60}")
    print(f"🔍 正在处理：{name} ({code}) - 任务9")
    print(f"{'='*60}")
    
    try:
        # 1. 读取数据（从task8）
        file_path = os.path.join(DATA_DIR, f"{code}_{name}_历史交易数据.csv")
        
        if not os.path.exists(file_path):
            print(f"❌ 错误：文件不存在 - {file_path}")
            # 尝试其他可能的文件名格式
            alternative_names = [
                f"{code}_{name}_历史数据.csv",
                f"{name}_{code}_历史交易数据.csv",
                f"{name}_历史交易数据.csv",
                f"{code}_历史交易数据.csv"
            ]
            
            found = False
            for alt_name in alternative_names:
                alt_path = os.path.join(DATA_DIR, alt_name)
                if os.path.exists(alt_path):
                    file_path = alt_path
                    print(f"✅ 找到替代文件: {alt_path}")
                    found = True
                    break
            
            if not found:
                print("   - ⚠️  尝试了所有可能的文件名格式，仍未找到文件")
                print("   - 💡 建议检查task8/stock_data目录中的文件命名格式")
                continue
        
        print(f"📄 读取文件: {file_path}")
        df = pd.read_csv(file_path)
        print(f"   - 原始数据量: {len(df)}行")
        
        # 2. 数据预处理
        print("🧹 数据预处理中...")
        df = preprocess_data(df)
        
        # 3. 整理日期格式
        print("📅 处理日期格式...")
        if 'date' not in df.columns:
            print("   - ⚠️  警告：未找到'date'列，尝试寻找其他日期列")
            date_cols = [col for col in df.columns if 'date' in col.lower() or 'time' in col.lower()]
            if date_cols:
                df.rename(columns={date_cols[0]: 'date'}, inplace=True)
                print(f"   - 使用列 '{date_cols[0]}' 作为日期列")
            else:
                print("   - ❌ 错误：未找到任何日期列")
                continue
        
        df["date"] = pd.to_datetime(df["date"])
        df.sort_values("date", inplace=True)
        df.set_index("date", inplace=True)
        
        # 4. 重命名列
        print("🔄 重命名列...")
        column_mapping = {
            "open": "Open",
            "high": "High", 
            "low": "Low",
            "close": "Close",
            "volume": "Volume"
        }
        # 智能列名匹配
        actual_columns = {col.lower(): col for col in df.columns}
        valid_columns = {}
        for orig_name, new_name in column_mapping.items():
            if orig_name.lower() in actual_columns:
                valid_columns[actual_columns[orig_name.lower()]] = new_name
            elif orig_name.capitalize() in df.columns:
                valid_columns[orig_name.capitalize()] = new_name
        
        if valid_columns:
            df.rename(columns=valid_columns, inplace=True)
            print(f"   - 重命名列: {valid_columns}")
        else:
            print("   - ⚠️  未找到需要重命名的列")
        
        # 5. 验证必需列
        print("✅ 验证数据列...")
        required_columns = ["Open", "High", "Low", "Close", "Volume"]
        missing_columns = [col for col in required_columns if col not in df.columns]
        if missing_columns:
            print(f"   - ❌ 缺失必需列: {missing_columns}")
            available_columns = list(df.columns)
            print(f"   - 可用列: {available_columns}")
            continue
        
        # 6. 确保数值列是float类型
        print("🔢 确保数值类型正确...")
        for col in ["Open", "High", "Low", "Close", "Volume"]:
            if col in df.columns:
                df[col] = df[col].astype(float)
                print(f"   - 列 '{col}' 数据类型: {df[col].dtype}")
        
        # 7. 计算技术指标
        print("🧮 计算技术指标...")
        df = calculate_technical_indicators(df)
        
        # 8. 展示结果
        print("\n📊 最近5天指标预览：")
        preview_cols = ["Close", "日收益率", "MA5_SMA", "MA10_SMA", "MA5_EMA", "MA10_EMA", "Volume"]
        available_preview_cols = [col for col in preview_cols if col in df.columns]
        if available_preview_cols:
            preview_df = df[available_preview_cols].tail().round(4)
            print(preview_df.to_string())
        else:
            print("   - 无可用预览列")
        
        # 9. 统计信息
        print("\n📈 数据统计信息：")
        print(f"   - 数据时间范围: {df.index.min().strftime('%Y-%m-%d')} 至 {df.index.max().strftime('%Y-%m-%d')}")
        print(f"   - 总交易日数: {len(df)}")
        print(f"   - 价格范围: {df['Low'].min():.2f} - {df['High'].max():.2f} 元")
        if '日收益率' in df.columns:
            print(f"   - 平均日收益率: {df['日收益率'].mean():.4f}%")
        
        # 10. 生成专业图表（保存到task9）
        print("\n🎨 生成专业K线图（保存到task9）...")
        chart_filename = f"{code}_{name}_任务9分析_{datetime.now().strftime('%Y%m%d_%H%M%S')}.png"
        chart_path = os.path.join(OUTPUT_DIR, chart_filename)
        
        create_professional_chart(df, name, code, chart_path)
        
        # 11. 保存处理后的数据（到task9）
        processed_filename = f"{code}_{name}_任务9处理后数据_{datetime.now().strftime('%Y%m%d')}.csv"
        processed_path = os.path.join(OUTPUT_DIR, processed_filename)
        df.to_csv(processed_path, encoding='utf_8_sig')
        print(f"💾 处理后数据已保存至（task9）: {processed_path}")
        
        # 12. 验证文件是否保存成功
        if os.path.exists(chart_path):
            print(f"✅ 验证：图表文件成功保存到task9")
        if os.path.exists(processed_path):
            print(f"✅ 验证：数据文件成功保存到task9")
        
    except Exception as e:
        print(f"❌ 处理 {name} ({code}) 时出错: {str(e)}")
        import traceback
        traceback.print_exc()
        continue

print("\n" + "="*70)
print("🎉 任务9股票分析完成！")
print(f"✅ 数据源: {DATA_DIR} (task8)")
print(f"✅ 输出目录: {OUTPUT_DIR} (task9)")
print("="*70)

# 最终验证
print("\n🔍 最终验证：检查task9输出目录内容")
if os.path.exists(OUTPUT_DIR):
    output_files = os.listdir(OUTPUT_DIR)
    if output_files:
        print(f"📁 task9/output_charts 目录包含 {len(output_files)} 个文件:")
        for file in output_files:
            print(f"   - {file}")
    else:
        print("⚠️  task9/output_charts 目录为空，可能没有成功生成文件")
else:
    print(f"❌ task9/output_charts 目录不存在，路径: {OUTPUT_DIR}")
```
